# **1. Insertion Sort**

```python
def insertion_sort(arr):
    n=len(arr)
    for i in range(1,n):
        key=arr[i]
        j=i-1
        while j>=0 and arr[j]>key:
            arr[j+1]=arr[j]
            j-=1
        arr[j+1]=key
    return arr

arr=[64,25,12,22,11]
print(insertion_sort(arr))
```

# **2. Bubble Sort**

```python
def bubble_sort(arr):
    n=len(arr)
    for i in range(n):
        for j in range(n-i-1):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr

arr=[5,3,8,4,2]
print(bubble_sort(arr))
```

# **3. Heap Sort**

```python
def heapify(ls,n,i):
    m=i
    l=2*i+1
    r=2*i+2
    if l<n and ls[l]>ls[m]:
        m=l
    if r<n and ls[r]>ls[m]:
        m=r
    if m!=i:
        ls[i],ls[m]=ls[m],ls[i]
        heapify(ls,n,m)

def heap_sort(ls):
    n=len(ls)
    for i in range(n//2-1,-1,-1):
        heapify(ls,n,i)
    for i in range(n-1,0,-1):
        ls[i],ls[0]=ls[0],ls[i]
        heapify(ls,i,0)
    return ls

ls=[15, -3, 0, 42, -8, 15, 4, -3]
res=heap_sort(ls)
print(res)
```

# **4. Selection Sort**

```python
def selection_sort(arr):
    n=len(arr)
    for i in range(n):
        min_idx=i
        for j in range(i+1,n):
            if arr[j]<arr[min_idx]:
                min_idx=j
        arr[i],arr[min_idx]=arr[min_idx],arr[i]
    return arr

arr=[64,25,12,22,11]
print(selection_sort(arr))
```

# **5. Merge Sort**

```python
def merge_sort(arr):
    if len(arr)<=1:
        return arr
    mid=len(arr)//2
    left=merge_sort(arr[:mid])
    right=merge_sort(arr[mid:])
    return merge(left,right)

def merge(left,right):
    result=[]
    i=0
    j=0
    while i<len(left) and j<len(right):
        if left[i]<=right[j]:
            result.append(left[i])
            i+=1
        else:
            result.append(right[j])
            j+=1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

arr=[38,27,43,3,9,82,10]
print(merge_sort(arr))
```

# **6. Quick Sort**

```python
def quick_sort(arr):
    if len(arr)<=1:
        return arr

    pivot=arr[-1]
    left=[]
    right=[]

    for x in arr[:-1]:
        if x<=pivot:
            left.append(x)
        else:
            right.append(x)

    return quick_sort(left)+[pivot]+quick_sort(right)

arr=[8,3,1,7,0,10,2]
print(quick_sort(arr))
```

# **7. KMP Search**

```python
def failure(P):
    F=[0]*len(P)
    k=0
    for i in range(1,len(P)):
        while k>0 and P[i]!=P[k]:
            k=F[k-1]
        if P[i]==P[k]:
            k+=1
        F[i]=k
    return F

def kmp(T,P):
    if len(P)==0:
        return []
    F=failure(P)
    result=[]
    j=0

    for i in range(len(T)):
        while j>0 and T[i]!=P[j]:
            j=F[j-1]
        if T[i]==P[j]:
            j+=1
        if j==len(P):
            result.append(i-j+1)
            j=F[j-1]
    return result

T="AABAACAADAABAABA"
P="AABA"
print(kmp(T,P))
```

# **8. Boyer-Moore Search**

```python
def boyer_moore(text,pattern):
    n=len(text)
    m=len(pattern)

    if m==0 or n<m:
        return []

    bad={}
    for i in range(m):
        bad[pattern[i]]=i

    result=[]
    i=0

    while i<=n-m:
        j=m-1
        while j>=0 and pattern[j]==text[i+j]:
            j-=1
        if j<0:
            result.append(i)
            i+=1
        else:
            char=text[i+j]
            if char in bad:
                i+=max(1,j-bad[char])
            else:
                i+=j+1

    return result

text="AABAACAADAABAABA"
pattern="AABA"
print(boyer_moore(text,pattern))
```
